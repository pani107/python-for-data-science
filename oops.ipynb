# ADD a method to class

```python
class Car:
    total_cars = 0 # count 
    def __init__(self , brand , model):
        self.brand = brand
        self.model = model
        Car.total_cars += 1
        
    def displayBrandModel(self):
        return f"{self.brand} {self.model}"
        
    def fuel_type(self):
        return "Petrol and Diesel"
        
car = Car("TATA" , "SIERRA ")
print(car.brand)
print(car.model)
print(car.displayBrandModel())
nexon = Car("TATA" , "NEXON")
print(Car.total_cars) # count 
        
```

# ENCAPSULATION

```python
class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner          # Public attribute
        self.__balance = balance    # Private attribute

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount

    def withdraw(self, amount):
        if 0 < amount <= self.__balance:
            self.__balance -= amount
        else:
            print("Insufficient funds")

    def get_balance(self):
        return self.__balance


# Usage
account = BankAccount("Alice", 1000)

account.deposit(500)
account.withdraw(200)

print("Balance:", account.get_balance())

# Direct access to private attribute is not allowed
# print(account.__balance)  # AttributeError
```

# Static Method

```python
class Car:
    def __init__(self , brand , model):
        self.brand = brand
        self.model = model
        
    @staticmethod   
    def info(): # not self
        return "the car have very attractive look"
        
car = Car("TATA" , "SIERRA")
print(car.brand)
print(car.model)
# car.model = "NEXON"
# print(car.model)
# print(car.info())
print(Car.info()) # without creating instance of class
# my_new_car = Car("TATA" , "SAFARI")
# print(my_new_car.model)
```

# Property Decorator

```python
class Car:
    def __init__(self, brand, model):
        self.brand = brand
        self._model = model  # private attribute

    @property
    def model(self):
        return self._model

    @model.setter
    def model(self, value):
        self._model = value

    @staticmethod
    def info():
        return "The car has a very attractive look"


car = Car("TATA", "SIERRA")
print(car.brand)
print(car.model)  # Calls the getter

car.model = "SAFARI"  # Calls the setter
print(car.model)

print(Car.info())
```
