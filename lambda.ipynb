```python
# sort list with first element.
nums = [(10,5),(20,3),(16,8),(25,7)]
nums.sort()
print(nums)
```

```python
# sort list with secound element.
nums = [(10,5),(20,3),(16,8),(25,7)]
nums.sort(key=lambda x: x[1])
print(nums)
```

```python
# print square of the list element by using lambda function.
nums = [2,3,4,5,6]
sq = list(map(lambda x:x**2,nums))
sq
```

```python
# use reduce function by using lambda 
from functools import reduce
nums = [2,3,4,6,7,8]
reduce(lambda a,b : a+b , nums)
```

```python
# find min value or element from the given list.
from functools import reduce
nums = [5,3,4,6,7,8]
min = reduce(lambda a,b: a if a < b else b ,nums)

# find max value or element from the given list.
from functools import reduce
nums = [5,3,4,6,7,8]
max = reduce(lambda a,b: a if a > b else b ,nums)
print(min)
print(max)
```
