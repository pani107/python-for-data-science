```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

compare = df.groupby('income')[['capital_gain', 'capital_loss']].mean().reset_index()

compare['income'] = compare['income'].replace({
    0: '<=50K',
    1: '>50K'
})

compare_melted = compare.melt(
    id_vars='income',
    value_vars=['capital_gain', 'capital_loss'],
    var_name='Metric',
    value_name='Average Value'
)

plt.figure(figsize=(8,5))
sns.barplot(data=compare_melted,
            x='income',
            y='Average Value',
            hue='Metric')

plt.title('Average Capital Gain and Capital Loss by Income')
plt.xlabel('Income Group')
plt.ylabel('Average Amount')
plt.show()
```

```python
#People earning >50K have much higher average capital gains.
#Capital gain shows a stronger difference between income groups than capital loss.
#Capital loss is relatively low for both groups.
#Investment-related earnings are more common among high-income individuals.
#Capital gain appears to be a stronger indicator of income level than capital loss.
```
