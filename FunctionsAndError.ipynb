# Abstraction

```python
from abc import abstractmethod, ABC
class Animal(ABC):
    @abstractmethod
    def speak(self):
        pass
class cat(Animal):
    def speak(self):
        print("i can meww")
class dog(Animal):
    def speak(self):
        print("i can bark")
c = cat()
c.speak()
```

# Error Handling

```python
class NegativeNumberError(Exception):
    """Custom Exception for negative numbers"""
    pass

try:
    print("=== Exception Handling Demo ===")

    num1 = int(input("Enter first number: "))
    num2 = int(input("Enter second number: "))

    if num1 < 0 or num2 < 0:
        raise NegativeNumberError("Negative numbers are not allowed.")

    result = num1 / num2

except ValueError:
    print("Error: Please enter valid integers only.")

except ZeroDivisionError:
    print("Error: Division by zero is not allowed.")

except NegativeNumberError as e:
    print("Custom Error:", e)

except Exception as e:
    print("Unexpected Error:", e)

else:
    print("Division Result =", result)

finally:
    print("Program execution completed.")
```

```python
import traceback
try:
    n = float(input("Enter a number:"))
    result = 12/n # ZeroDivisionError
    with open("xyz.txt") as f: # FileNotFoundError
        print(f.read())
except ZeroDivisionError as e:
    print(f"Error occurred:", e)
    print(traceback.print_exc())
except FileNotFoundError as e:
    print(f"Error occurred:", e)
    print(traceback.print_exc())
except Exception as e: # general exception; other than above handles ones
    print("Undefined error:", e)
    print(traceback.print_exc())


print("Application executed successfully!")
```

# function logger and wrapper

```python
def add(a , b):
    print(f"{a} + {b} = {a + b}")
add(4,5)
```

```python
def logger(fun):
    def wrapper():
        print(f"befor calling the function {fun.__name__}")
        fun()
        print(f"After calling the function{fun.__name__}")
    return wrapper

@logger
def greet():
    print("hello world")
greet()
# @logger
# def welcome():
#     print("welcome world")
#welcome()
```

```python
def add(*args):
    return sum(args)

def multiply(*args):
    result = 1
    for num in args:
        result *= num
    return result

def divide(*args):
    if len(args) == 0:
        return "No numbers provided"

    result = args[0]
    for num in args[1:]:
        if num == 0:
            return "Division by zero is not allowed"
        result /= num
    return result


print("Addition:")
print(add(10, 20))
print(add(10, 20, 30))
print(add())

print("\nMultiplication:")
print(multiply(10, 20))
print(multiply(10, 20, 30))
print(multiply())

print("\nDivision:")
print(divide(100, 2))
print(divide(100, 2, 5))
print(divide(100, 0))
```

```python
#variable argument.
def calc(*args, **kwargs):
    op = kwargs.get('operator')

    if op == '+':
        return sum(args)

    elif op == '*':
        result = 1
        for num in args:
            result *= num
        return result

    else:
        print(f"Wrong operator: {op}")


print(calc(10, 20, operator="+"))
print(calc(10, 2, 3, operator="*"))
```
