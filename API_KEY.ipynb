```python
import os
from pathlib import Path
from google.colab import userdata
from dotenv import load_dotenv
from openai import OpenAI

OPENAI_API_KEY = userdata.get('OPENAI_API_KEY').strip()
client = OpenAI(api_key=OPENAI_API_KEY)
MODEL = "gpt-4o-mini"
```

```python
 # Helper to call the chat API with optional system prompt and temperature.
def chat(prompt: str, system: str = None, temperature: float = 0.7, max_tokens: int = 300) -> str:
    messages = []
    if system:
        messages.append({"role": "system", "content": system})
    messages.append({"role": "user", "content": prompt})


    response = client.chat.completions.create(
        model=MODEL,
        messages=messages,
        temperature=temperature,
        max_tokens=max_tokens,
    )
    return response.choices[0].message.content.strip()
```

```python
# Zero-shot: classify book description as Fiction or Nonfiction.
book_desc = (
    "A memoir of a scientist who spent decades researching climate change "
    "and testified before government committees."
)


zero_shot_prompt = f"""
Classify the following book description as either Fiction or Nonfiction.
Reply with only one word.


Description: {book_desc}
"""


print(chat(zero_shot_prompt, temperature=1))
```

```python
# Role prompt: librarian persona recommends books in a specific style.
system_role = (
    "You are a warm, knowledgeable children's librarian. "
    "Recommend books in 2-3 sentences. Mention age range and one reason to read."
)


user_query = "My 8-year-old loves animals and adventure stories."


print(chat(user_query, system=system_role, temperature=1.5))
```

```python
few_shot_prompt = """
Label each book description with one emotional tone: Happy, Sad, Suspenseful, or Neutral.


Description: A lonely robot learns what friendship means on a distant planet.
Tone: Happy


Description: A detective follows clues through foggy streets as danger closes in.
Tone: Suspenseful


Description: A historian documents the fall of an ancient empire.
Tone: Neutral


Description: A family says goodbye as their grandmother's health fades.
Tone: Sad


Description: A young wizard discovers a hidden library beneath the school.
Tone:
"""


print(chat(few_shot_prompt, temperature=2))
```
